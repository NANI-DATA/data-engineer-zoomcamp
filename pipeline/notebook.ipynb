```python
from sqlalchemy import create_engine
engine = create_engine('postgresql://root:root@localhost:5432/ny_taxi')
```

```python
print(pd.io.sql.get_schema(
    df1,
    name='green_tripdata_2025_11',
    con=engine
))
```
